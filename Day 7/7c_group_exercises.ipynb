## Exercise 2: Comparing Bayesian Prior Distributions

**Estimated Time:** 8 minutes

### Background

A prior distribution represents uncertainty about an unknown parameter before new observations are collected.

The Beta distribution is commonly used to represent uncertainty about a probability.

$$
p \sim \operatorname{Beta}(\alpha,\beta)
$$

The mean of a Beta distribution is:

$$
E[p] = \frac{\alpha}{\alpha+\beta}
$$

Different values of $\alpha$ and $\beta$ represent different prior assumptions.

### Scenario
