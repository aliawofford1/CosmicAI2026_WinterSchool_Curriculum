#Exercise 3

Objective: Write a Python function that reads the poem Annabel Lee by Edgar Allan Poe from poem.txt and identifies all words containing between 2 and 5 characters.

Create a function that opens and reads the file poem.txt using with open().
Read the file line by line using a for loop.
Separate each line into individual words and remove any unnecessary whitespace or punctuation.
Identify all words containing between 2 and 5 characters (inclusive).
Store the identified words in a list.
Display the list of words and the total number of words identified.
Write the identified words to a new text file named filtered_words.txt, with each word appearing on a separate line.

```python

def find_words():

    # Create an empty list to store the words
    words_list = []

    # Open the poem file
    with open("poem.txt", "r") as file:

        # Read the file line by line
        for line in file:

            # Separate each line into words
            words = line.split()

            # Check each word
            for word in words:

                # Remove punctuation and whitespace
                word = word.strip(".,;:!?()— \n")

                # Check the length of the word
                if len(word) >= 2 and len(word) <= 5:

                    # Add the word to the list
                    words_list.append(word)

    # Display the words
    print("Words containing 2 to 5 characters:")
    print(words_list)

    # Display the total number of words
    print("Total words:", len(words_list))

    # Write the words to a new file
    with open("filtered_words.txt", "w") as file:

        for word in words_list:
            file.write(word + "\n")


# Call the function
find_words()
```

# Exercise 4

**Objective:** Write a Python function that reads the poem *Annabel Lee* from `poem.txt` and determines how many times the word "sea" appears.

1. Create a function that opens and reads the file `poem.txt` using `with open()`.
2. Use a `for` loop to read each line and separate it into individual words.
3. Remove unnecessary whitespace and punctuation, and convert words to lowercase to ensure accurate counting.
4. Count how many times the word "sea" appears throughout the poem.
5. Display the total number of occurrences of the word "sea".
6. Write the result to a new file named `word_count.txt`.
7. Use append mode (`'a'`) to add the total number of lines in the poem to `word_count.txt` without overwriting the existing contents.


```python

def count_sea():

    # Create variables to store the counts
    sea_count = 0
    line_count = 0

    # Open the poem file
    with open("poem.txt", "r") as file:

        # Read the file line by line
        for line in file:

            # Count each line
            line_count = line_count + 1

            # Separate the line into words
            words = line.split()

            # Check each word
            for word in words:

                # Remove punctuation and convert to lowercase
                word = word.lower()
                word = word.strip(".,;:!?()— \n")

                # Check if the word is sea
                if word == "sea":

                    # Increase the counter
                    sea_count = sea_count + 1

    # Display the results
    print("The word 'sea' appears", sea_count, "times.")

    # Write the word count to a new file
    with open("word_count.txt", "w") as file:
        file.write("Sea count: " + str(sea_count) + "\n")

    # Append the line count to the same file
    with open("word_count.txt", "a") as file:
        file.write("Total lines: " + str(line_count) + "\n")

    print("Total lines:", line_count)


# Call the function
count_sea()
```
