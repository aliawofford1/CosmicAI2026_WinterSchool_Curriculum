### Introduction to Bayesian Statistics

Bayesian statistics is a statistical approach that allows us to update
our knowledge or beliefs about an event as new information becomes available.

Unlike frequentist statistics, which interprets probability in terms of
long-run frequencies, Bayesian statistics uses probability to represent
uncertainty about unknown quantities.

Bayesian inference combines three important concepts:

- **Prior Probability:** What we initially believe or know about an
  event or hypothesis before observing new data.
- **Likelihood:** How probable the observed data are under a
  particular hypothesis.
- **Posterior Probability:** Our updated probability after incorporating
  the new evidence.

This process allows us to continually refine our understanding
as additional information becomes available.

#### Where Is Bayesian Statistics Used?

Bayesian methods are widely used in:

- **Scientific Research:** Incorporating prior scientific knowledge
  into statistical models.
- **Medical Diagnosis:** Updating the probability of a disease
  based on diagnostic test results.
- **Weather Forecasting:** Updating predictions using new
  atmospheric observations.
- **Machine Learning:** Quantifying uncertainty in model parameters
  and predictions.
- **Neural Networks:** Using probability distributions to represent
  uncertainty in learned model parameters.

Bayesian statistics is especially valuable when understanding
uncertainty is just as important as making an accurate prediction.

### Frequentist vs. Bayesian Statistics

Frequentist and Bayesian statistics are two approaches to
statistical inference.

Although both use observed data to draw conclusions, they differ
in how they interpret probability and uncertainty.

| Feature | Frequentist | Bayesian |
|---|---|---|
| Probability | Long-run frequency of events | Represents uncertainty about events or unknown quantities |
| Parameters | Fixed but unknown | Represented by probability distributions |
| Prior Knowledge | Does not explicitly use prior distributions | Incorporates prior information |
| Statistical Inference | Confidence intervals and hypothesis tests | Posterior distributions and credible intervals |
| New Data | Analysis can be repeated or extended | Posterior probabilities can be updated |
| Machine Learning | Conventional regression and neural networks | Bayesian regression and Bayesian neural networks |

#### Example: Predicting Rain

Suppose we want to estimate the probability of rain tomorrow.

**Frequentist Approach:**

We could analyze historical weather observations and calculate
how frequently rain occurred under similar atmospheric conditions.

**Bayesian Approach:**

We could begin with an initial probability of rain based on
historical observations and update that probability using
new atmospheric measurements.

Both approaches can produce useful predictions.

The important distinction is that Bayesian inference explicitly
combines prior information with new evidence to update uncertainty.

**Key Takeaway:** Frequentist statistics uses repeated-sampling
principles for statistical inference, while Bayesian statistics
provides a framework for updating probability distributions
as new information becomes available.

### Bayes' Theorem

...allows us to find $P(\text{x}|\text{y})$ when we have $P(\text{y}|\text{x})$:
$$ P(\text{x}|\text{y}) = \frac{P(\text{x})P(\text{y}|\text{x})}{P(\text{y})} $$

Let's use the *xkcd* [exploding sun cartoon](https://xkcd.com/1132/) as an example. Using a standard $\alpha = .05$ threshold, the frequentist rejected the null hypothesis that the sun hadn't exploded because the probability the neutrino detector outputs `YES` when the sun hasn't exploded is $\frac{1}{6} \times \frac{1}{6} = \frac{1}{36} \approx 0.0278$, which is $<.05$.

Using Bayes' theorem, the Bayesian statistician incorporates additional information -- largely related to the probability that the sun has exploded irrespective what the neutrino detector says -- to draw a different conclusion.

Let's likewise use some back-of-the-envelope figures to estimate the probability the sun has exploded ($x = \text{exploded}$) given the neutrino detector output `YES` ($y = \text{YES}$); that is, $P(x|y)$. To find this, we'll need $P(y|x)$, $P(x)$, and $P(y)$.

$P(y = \text{YES} | x = \text{exploded}) = \frac{35}{36} \approx 0.972 $.

$P(x = \text{exploded})$ can be roughly estimated. It is generous to assume a $\frac{1}{1000}$ chance because for every thousand days that pass, the sun explodes far less often than once. Further, unless the sun had exploded only in the past few minutes, we'd already be dead. So, the probability that we are alive in an instant where the sun has exploded is extremely small. Anyway, let's go with $\frac{1}{1000}$ because even with this exceedingly generous figure, we'll demonstrate the point.

$P(y = \text{YES})$: As is often the case, this probability in the Bayes' theorem denominator can be calculated with information we already have because:
$$ P(\text{y}) = \sum_x P(\text{y}|x)P(x) $$
Summing over the two possible states of x ($x =$ exploded, $x =$ not exploded):
$$ P(y = \text{YES}) = P(\text{YES given exploded})P(\text{exploded}) + P(\text{YES given not exploded})P(\text{not exploded}) $$
$$ = \left(\frac{35}{36}\right)\left(\frac{1}{1000}\right) + \left(\frac{1}{36}\right)\left(\frac{999}{1000}\right) $$
...which comes out to $P(y = \text{YES}) \approx 0.0287$:

```python
py = (35/36.)*(1/1000.) + (1/36.)*(999/1000.)
py
```

Now we have everything we need to apply Bayes' theorem:

```python
py_givenx = 35/36
px = 1/1000
```

```python
def bayes(my_px, my_pygivenx, my_py):
    return (my_px*my_pygivenx)/my_py
```

```python
bayes(px, py_givenx, py)
```

### Interpreting the Results

Using Bayes' Theorem, we find that the probability of the sun having exploded, given that the detector reports "YES," is approximately **3.38%**.

Although the detector is highly accurate, the probability that the sun has actually exploded remains relatively low.

This is because the event itself is extremely unlikely, with a prior probability of only 0.1%.

**Key Takeaway:** Bayesian inference demonstrates how prior probabilities influence our interpretation of new evidence. Even when a test or detector is highly accurate, a positive result does not necessarily indicate that an unlikely event has occurred.

### Hands-On Tutorial: Bayesian Inference with Coin Flips

In this exercise, we will explore how Bayesian statistics allows us to
update our understanding of an unknown quantity as new data become available.

Suppose we have a coin, but we do not know whether it is fair.

A fair coin has a 50% probability of landing heads. However, we want
to determine whether our coin is fair by observing the results of
several coin flips.

Using Bayesian inference, we will:

1. Define an initial belief about the probability of heads.
2. Simulate coin flips using Python.
3. Update our belief using the observed results.
4. Visualize how our uncertainty changes as more data become available.

**Learning Objective:** Understand how prior information and observed
data combine to produce an updated probability distribution.

We will use NumPy for numerical calculations and Matplotlib
for data visualization.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

# Set a random seed for reproducibility
np.random.seed(42)
```

#### Step 1: Define Our Prior Belief

Before flipping the coin, we need to establish our initial belief
about the probability of heads.

We will use a Beta distribution to represent this uncertainty.

The Beta distribution has two parameters:

- **Alpha (α):** Related to the number of observed heads.
- **Beta (β):** Related to the number of observed tails.

We will begin with α = 1 and β = 1.

This produces a uniform distribution, meaning we initially consider
all possible probabilities of heads between 0 and 1 equally plausible.

```python
# Define the prior distribution
alpha_prior = 1
beta_prior = 1

# Possible values for the probability of heads
p = np.linspace(0, 1, 500)

# Calculate the prior probability density
prior = beta.pdf(p, alpha_prior, beta_prior)

plt.figure(figsize=(8, 4))
plt.plot(p, prior, label="Prior Distribution")
plt.xlabel("Probability of Heads")
plt.ylabel("Probability Density")
plt.title("Prior Belief About Coin Fairness")
plt.legend()
plt.ylim(0,1.5)
plt.grid(alpha=0.3)
plt.show()
```

Expected result: A horizontal line indicating that all probabilities between 0 and 1 are initially equally plausible.

#### Step 2: Collect Observations

Now we will simulate 10 coin flips.

For this example, we will use a coin with a 70% probability
of landing heads.

However, imagine that we do not know this probability beforehand.

Our goal is to use the observed results to estimate it.

```python
# Unknown probability of heads (used to generate simulated data)
true_probability = 0.7

# Simulate 10 coin flips
flips = np.random.binomial(1, true_probability, size=10)

# Count heads and tails
heads = np.sum(flips)
tails = len(flips) - heads

print("Coin flips:", flips)
print("Number of heads:", heads)
print("Number of tails:", tails)
```

#### Step 3: Update Our Belief

After observing the coin flips, we can update our prior distribution.

Because we are using a Beta prior with Bernoulli observations,
the posterior distribution is also a Beta distribution.

The updated parameters are:

α_posterior = α_prior + number of heads

β_posterior = β_prior + number of tails

The posterior distribution combines our prior belief
with the evidence collected from the coin flips.

```python
# Update the parameters
alpha_post = alpha_prior + heads
beta_post = beta_prior + tails

# Calculate posterior distribution
posterior = beta.pdf(p, alpha_post, beta_post)

# Calculate posterior mean
posterior_mean = alpha_post / (alpha_post + beta_post)

print(f"Posterior mean probability of heads: {posterior_mean:.3f}")
```

```python
plt.figure(figsize=(8, 4))

plt.plot(p, prior, label="Prior", linestyle="--")
plt.plot(p, posterior, label="Posterior")

plt.axvline(
    true_probability,
    color="black",
    linestyle=":",
    label="True Probability"
)

plt.xlabel("Probability of Heads")
plt.ylabel("Probability Density")
plt.title("Bayesian Updating: Prior vs. Posterior")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

#### Interpreting the Results

The figure compares our initial belief (prior distribution) with our updated belief (posterior distribution) after observing 10 coin flips.

**Prior Distribution (Dashed Line):**

Before collecting data, we assumed that all probabilities of heads between 0 and 1 were equally plausible.

**Posterior Distribution (Solid Curve):**

After observing 6 heads and 4 tails, our posterior distribution becomes more concentrated around probabilities consistent with the observations.

The posterior mean is approximately **0.583 (58.3%)**, compared with the true probability of 0.7 (70%) used to generate the data.

**True Probability (Dotted Vertical Line):**

The dotted line represents the actual probability of heads used in our simulation.

Although our posterior estimate differs from the true probability, collecting additional observations will generally help us estimate the underlying probability more precisely.

**Key Takeaway:** Bayesian inference allows us to update our understanding of an unknown parameter as new data become available. As we collect more observations, uncertainty generally decreases, and our posterior distribution becomes more concentrated.

### Connecting Bayesian Inference to Machine Learning

In our coin-flipping example, we used Bayesian inference to estimate
an unknown probability.

Bayesian machine learning applies similar principles to estimating
unknown model parameters.

For example, conventional neural networks typically learn a single
set of weights and biases during training.

Bayesian neural networks instead represent uncertainty in model
parameters using probability distributions.

This can help researchers understand uncertainty in model predictions,
which is especially important in scientific applications.
