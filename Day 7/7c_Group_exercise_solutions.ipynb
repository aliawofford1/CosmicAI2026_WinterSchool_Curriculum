## Indvidual Exercise 1:

```python
# Given probabilities
prior_event = 0.01
sensitivity = 0.95
false_positive_rate = 0.05

# TASK 1: Total probability of a positive detection
p_positive = (
    sensitivity * prior_event
    + false_positive_rate * (1 - prior_event)
)

# TASK 2: Posterior probability
posterior_event = (
    sensitivity * prior_event
) / p_positive

print(f"Posterior Probability (Prior = 0.01): {posterior_event:.4f}")

# TASK 3: Update prior
new_prior = 0.10

new_p_positive = (
    sensitivity * new_prior
    + false_positive_rate * (1 - new_prior)
)

new_posterior = (
    sensitivity * new_prior
) / new_p_positive

# TASK 4: Compare results
print(f"Posterior Probability (Prior = 0.10): {new_posterior:.4f}")
```

## Reflection Solutions

#### 1. Why is the posterior probability substantially lower than the detector's sensitivity when the event is rare?

**Answer:** Detector sensitivity measures the probability of a positive detection when an event actually occurs.

The posterior probability measures the probability that an event occurred given a positive detection.

When an event is rare, most observations do not contain the event. Even a relatively small false-positive rate can therefore produce many false detections.

#### 2. How does increasing the prior probability affect the posterior probability?

**Answer:** Increasing the prior probability increases the posterior probability when the detector's sensitivity and false-positive rate remain unchanged.

A positive detection provides stronger evidence for an event when the event was already considered more likely.

### Key Takeaway

Bayesian inference combines prior probabilities with new evidence.

The reliability of a positive detection depends on both the detector's performance and the underlying frequency of the event.

## Indvidual Exercise 2:






```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

p = np.linspace(0.001, 0.999, 500)

priors = {
    "Beta(1,1)": (1, 1),
    "Beta(5,5)": (5, 5),
    "Beta(8,2)": (8, 2)
}

plt.figure(figsize=(8, 5))

for name, (alpha, beta_param) in priors.items():

    # TASK 1: Calculate prior mean
    prior_mean = alpha / (alpha + beta_param)

    print(f"{name}: Mean = {prior_mean:.3f}")

    # TASK 2: Calculate probability density
    density = beta.pdf(p, alpha, beta_param)

    plt.plot(
        p,
        density,
        label=name
    )

plt.xlabel("Probability of Heads")
plt.ylabel("Probability Density")
plt.title("Comparison of Bayesian Prior Distributions")

plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

###  Reflection Solutions

#### 1. Why do Beta(1,1) and Beta(5,5) have the same mean but different shapes?

**Answer:** Both distributions are symmetric around 0.5.

However, Beta(5,5) is more concentrated around 0.5, indicating stronger prior confidence that the probability of heads is close to 0.5.

Beta(1,1) is uniform and represents greater uncertainty across the possible probabilities.

#### 2. How might choosing different priors affect the posterior distribution when only a small number of observations is available?

**Answer:** When limited data are available, the prior distribution can substantially influence the posterior.

A strongly concentrated prior may require more observations before the posterior shifts significantly away from the prior expectation.

As more observations are collected, the likelihood generally becomes more influential.

### Key Takeaway

Prior distributions describe uncertainty before observing new data.

Two priors can have the same mean while representing different levels of uncertainty.

## Individual Exercise 3:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

np.random.seed(42)

true_probability = 0.7
n_flips = 100

alpha_prior = 1
beta_prior = 1

# TASK 1: Simulate coin flips
flips = np.random.binomial(
    1,
    true_probability,
    size=n_flips
)

# TASK 2: Prior distribution
p = np.linspace(0.001, 0.999, 500)

prior_density = beta.pdf(
    p,
    alpha_prior,
    beta_prior
)

plt.figure(figsize=(9, 5))

plt.plot(
    p,
    prior_density,
    linestyle="--",
    label="Prior: Beta(1,1)"
)

sample_sizes = [10, 30, 100]

for n in sample_sizes:

    # TASK 3: Count observations
    heads = np.sum(flips[:n])
    tails = n - heads

    # Posterior parameters
    alpha_post = alpha_prior + heads
    beta_post = beta_prior + tails

    # TASK 4: Posterior mean
    posterior_mean = (
        alpha_post / (alpha_post + beta_post)
    )

    print(
        f"{n} Flips: "
        f"Heads = {heads}, "
        f"Tails = {tails}, "
        f"Posterior Mean = {posterior_mean:.3f}"
    )

    # TASK 5: Posterior distribution
    posterior_density = beta.pdf(
        p,
        alpha_post,
        beta_post
    )

    plt.plot(
        p,
        posterior_density,
        label=f"Posterior After {n} Flips"
    )

# TASK 6: Compare with true probability
plt.axvline(
    true_probability,
    color="black",
    linestyle=":",
    label="True Probability = 0.7"
)

plt.xlabel("Probability of Heads")
plt.ylabel("Probability Density")
plt.title("Sequential Bayesian Updating")

plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### Reflection Solutions

#### 1. How does the posterior distribution change as more observations are collected?

**Answer:** As more observations are collected, the posterior distribution generally becomes more concentrated around probabilities supported by the data.

The narrowing distribution indicates reduced uncertainty about the unknown probability of heads.

#### 2. Why might the posterior mean differ from the true probability after only 10 observations?

**Answer:** A small sample may not accurately represent the underlying probability because of random sampling variability.

The prior also influences the posterior mean.

As additional observations are collected, the estimate generally becomes more stable.

### Key Takeaway

Sequential Bayesian updating incorporates new observations into an existing probability distribution.

Increasing the amount of data generally improves the precision of posterior estimates.

## Individual Exercise 4:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

successes = 18
failures = 7

alpha_prior = 1
beta_prior = 1

# TASK 1: Posterior parameters
alpha_post = alpha_prior + successes
beta_post = beta_prior + failures

# TASK 2: Posterior mean
posterior_mean = (
    alpha_post / (alpha_post + beta_post)
)

# TASK 3: 95% credible interval
lower = beta.ppf(
    0.025,
    alpha_post,
    beta_post
)

upper = beta.ppf(
    0.975,
    alpha_post,
    beta_post
)

print(f"Posterior Mean: {posterior_mean:.4f}")
print(f"95% Credible Interval: ({lower:.4f}, {upper:.4f})")

# TASK 4: Posterior density
p = np.linspace(0.001, 0.999, 500)

posterior_density = beta.pdf(
    p,
    alpha_post,
    beta_post
)

plt.figure(figsize=(9, 5))

plt.plot(
    p,
    posterior_density,
    label="Posterior Distribution"
)

# TASK 5: Shade credible interval
plt.fill_between(
    p,
    posterior_density,
    where=(p >= lower) & (p <= upper),
    alpha=0.3,
    label="95% Credible Interval"
)

plt.axvline(
    posterior_mean,
    color="black",
    linestyle="--",
    label="Posterior Mean"
)

plt.xlabel("Detection Probability")
plt.ylabel("Probability Density")
plt.title("Posterior Distribution and Credible Interval")

plt.legend()
plt.grid(alpha=0.3)
plt.show()

# TASK 6: Probability detection rate exceeds 0.5
prob_above_half = 1 - beta.cdf(
    0.5,
    alpha_post,
    beta_post
)

print(f"P(p > 0.5 | data): {prob_above_half:.4f}")
```

### Reflection Solutions

#### 1. How does a Bayesian credible interval differ from a frequentist confidence interval?

**Answer:** A Bayesian credible interval represents a range containing a specified proportion of the posterior probability.

Under the assumed prior, likelihood, and observed data, a 95% credible interval contains 95% of the posterior probability for the unknown parameter.

A frequentist confidence interval instead describes a procedure that achieves a specified coverage rate across repeated samples.

#### 2. What does P(p > 0.5 | data) tell us?

**Answer:** It represents the posterior probability that the unknown detection probability exceeds 0.5.

This allows researchers to make a probability statement directly about the unknown parameter under the assumed Bayesian model.

### Key Takeaway

Posterior distributions quantify uncertainty about unknown parameters.

Credible intervals and posterior probabilities provide useful summaries of that uncertainty.

## Individual Exercise 5:

```python
import numpy as np
from scipy.stats import beta

np.random.seed(42)

successes = 12
failures = 8

alpha_prior = 1
beta_prior = 1

# TASK 1: Posterior parameters
alpha_post = alpha_prior + successes
beta_post = beta_prior + failures

# TASK 2: Posterior predictive probability
predictive_probability = (
    alpha_post / (alpha_post + beta_post)
)

# TASK 3: Generate posterior samples
posterior_samples = beta.rvs(
    alpha_post,
    beta_post,
    size=10000
)

# TASK 4: Simulate future observations
future_observations = np.random.binomial(
    1,
    posterior_samples
)

# TASK 5: Calculate simulated success rate
simulated_probability = np.mean(future_observations)

# TASK 6: Compare results
print(
    f"Theoretical Predictive Probability: "
    f"{predictive_probability:.4f}"
)

print(
    f"Simulated Predictive Probability: "
    f"{simulated_probability:.4f}"
)
```

###  Reflection Solutions

#### 1. Why is the posterior predictive probability equal to the posterior mean in this example?

**Answer:** In the Beta-Bernoulli model, the probability of success in the next observation is obtained by averaging the unknown success probability over its posterior distribution.

Therefore:

$$
P(Y_{\text{next}}=1|\text{data})
=
E[p|\text{data}]
$$

This relationship applies to a single future Bernoulli observation under the assumed model.

#### 2. How does Bayesian prediction incorporate uncertainty about the unknown parameter?

**Answer:** Bayesian prediction averages over possible parameter values according to their posterior probabilities.

Rather than assuming a single fixed estimated probability, the posterior predictive calculation incorporates uncertainty about the unknown parameter.

### Key Takeaway

Bayesian inference can be used for both parameter estimation and prediction.

Posterior predictive methods account for uncertainty in unknown parameters when estimating future outcomes.
