## Group Exercise 1:

```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt

n = 3
p = 0.5

x = np.arange(0, n + 1)

# Calculate PMF
probabilities = st.binom.pmf(x, n, p)

# Verify total probability
total_probability = np.sum(probabilities)

# Expected value
expected_value = np.sum(x * probabilities)

print("Number of detections:", x)
print("Probabilities:", probabilities)
print(f"Total probability: {total_probability:.2f}")
print(f"Expected value: {expected_value:.2f}")

# PMF visualization
plt.figure(figsize=(7, 4))

plt.bar(x, probabilities)

plt.xlabel("Number of Successful Detections")
plt.ylabel("Probability")
plt.title("Binomial Probability Mass Function")

plt.xticks(x)
plt.show()
```

### Instructor Reflection Solutions

#### 1. Why is a PMF appropriate for this random variable?

**Answer:** The number of successful detections is a discrete random variable because it can only take integer values between 0 and 3.

Therefore, a probability mass function (PMF) is appropriate because it assigns a probability to each possible outcome.

#### 2. What does the expected value represent in this scenario?

**Answer:** The expected value of 1.5 represents the average number of successful detections expected across many repeated sets of three observations.

The expected value is calculated using:

$$
E[X] = \sum_x xP(X=x)
$$

For this experiment:

$$
E[X] = (0)(0.125) + (1)(0.375) + (2)(0.375) + (3)(0.125)
$$

$$
E[X] = 1.5
$$

#### 3. Would the expected value necessarily equal the number of detections in any single experiment?

**Answer:** No. A single experiment can produce only 0, 1, 2, or 3 detections.

The expected value of 1.5 does not represent a possible outcome of one experiment. Instead, it describes the long-run average across many repeated experiments.

#### 4. How would the PMF change if the detection probability increased to 0.8?

**Answer:** If the detection probability increases from 0.5 to 0.8, the distribution shifts toward higher numbers of successful detections.

The expected value would increase from:

$$
E[X] = 3(0.5) = 1.5
$$

to:

$$
E[X] = 3(0.8) = 2.4
$$

Three successful detections would become the most likely outcome.

---

### Key Takeaway

A **probability mass function (PMF)** describes the probability of each possible outcome for a discrete random variable.

The **expected value** represents the probability-weighted average outcome over many repeated experiments.

```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt

environment_A = np.array([
    18, 19, 20, 21, 20,
    22, 21, 19, 20, 21,
    22, 23, 20, 19, 21
])

environment_B = np.array([
    15, 17, 19, 20, 18,
    22, 24, 21, 23, 25,
    24, 26, 22, 20, 23
])

for name, data in [
    ("Environment A", environment_A),
    ("Environment B", environment_B)
]:
    print(f"\n{name}")

    print(f"Mean: {np.mean(data):.2f}")
    print(f"Median: {np.median(data):.2f}")
    print(f"Mode: {st.mode(data).mode}")

    q1, q2, q3 = np.percentile(data, [25, 50, 75])

    print(f"Q1: {q1:.2f}")
    print(f"Q2: {q2:.2f}")
    print(f"Q3: {q3:.2f}")

    print(f"Variance: {np.var(data):.2f}")
    print(f"Standard Deviation: {np.std(data):.2f}")
    print(f"Standard Error: {st.sem(data, ddof=0):.2f}")

# Boxplots
plt.figure(figsize=(8, 5))

plt.boxplot(
    [environment_A, environment_B],
    tick_labels=["Environment A", "Environment B"]
)

plt.ylabel("Temperature (°C)")
plt.title("Temperature Distribution Comparison")

plt.show()

# Sample covariance
covariance = np.cov(
    environment_A,
    environment_B,
    ddof=1
)[0, 1]

# Pearson correlation
correlation, p_value = st.pearsonr(
    environment_A,
    environment_B
)

print(f"\nSample Covariance: {covariance:.3f}")
print(f"Pearson Correlation: {correlation:.3f}")
```

```python
true_positive = 70
false_positive = 30
false_negative = 20
true_negative = 80

total = 200

# Marginal probabilities
p_signal = (true_positive + false_negative) / total

p_predict_signal = (
    true_positive + false_positive
) / total

# Conditional probabilities
p_signal_given_prediction = (
    true_positive / (true_positive + false_positive)
)

p_prediction_given_signal = (
    true_positive / (true_positive + false_negative)
)

# Joint probability
p_joint = true_positive / total

# Independence comparison
p_independent = p_signal * p_predict_signal

print(f"P(Signal): {p_signal:.3f}")
print(f"P(Predict Signal): {p_predict_signal:.3f}")

print(
    f"P(Signal | Predict Signal): "
    f"{p_signal_given_prediction:.3f}"
)

print(
    f"P(Predict Signal | Signal): "
    f"{p_prediction_given_signal:.3f}"
)

print(f"P(Signal and Predict Signal): {p_joint:.3f}")
print(f"P(Signal) × P(Predict Signal): {p_independent:.3f}")

if abs(p_joint - p_independent) < 1e-10:
    print("The events are independent.")
else:
    print("The events are not independent.")
```

### Instructor Reflection Solutions

#### 1. What is the difference between marginal and conditional probability?

**Answer:** Marginal probability describes the likelihood of an event occurring without considering additional conditions.

Conditional probability describes the likelihood of an event occurring given that another event has already occurred.

For example:

- **Marginal probability:** The probability that a signal is present is 0.450.
- **Conditional probability:** The probability that a signal is present given that the model predicts a signal is 0.700.

#### 2. Why are the two conditional probabilities calculated in Questions 3 and 4 different?

**Answer:** The conditional probabilities differ because they use different reference groups.

- $P(\text{Signal} \mid \text{Predict Signal}) = 0.700$ represents the proportion of predicted signals that are actually present.
- $P(\text{Predict Signal} \mid \text{Signal}) = 0.778$ represents the proportion of actual signals that the model correctly identifies.

These probabilities are different because the conditions change the group of observations being considered.

#### 3. Are the events independent? Explain.

**Answer:** No, the events are not independent.

Two events are independent if:

$$
P(A \cap B) = P(A)P(B)
$$

In this exercise:

$$
P(A \cap B) = 0.350
$$

However:

$$
P(A)P(B) = 0.450 \times 0.500 = 0.225
$$

Since:

$$
0.350 \neq 0.225
$$

**The events are not independent.**

This indicates that the model's prediction is statistically associated with whether a signal is actually present.

#### 4. What additional information would you need to evaluate conditional independence between two events?

**Answer:** To evaluate conditional independence, we would need information about a third variable, such as observation quality or signal strength.

For example, we could examine whether the model's predictions and actual signal presence are independent after accounting for observation quality.

Conditional independence requires:

$$
P(A \cap B \mid C) = P(A \mid C)P(B \mid C)
$$

where:

- $A$ represents the event that a signal is present.
- $B$ represents the event that the model predicts a signal.
- $C$ represents an additional condition, such as observation quality.

To test conditional independence, we would need sufficient observations to calculate these probabilities for each relevant value or category of $C$.

### Key Takeaway

**Marginal probability** describes the likelihood of an event without additional conditions, while **conditional probability** describes the likelihood of an event given another event.

Two events are **independent** when knowing that one occurred does not change the probability of the other.

**Conditional independence** occurs when two events are independent after accounting for a third variable.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Generate population
population = np.random.uniform(0, 1, 10000)

# TASK 1: Calculate population mean and standard deviation
population_mean = np.mean(population)
population_std = np.std(population)

print(f"Population Mean: {population_mean:.4f}")
print(f"Population Standard Deviation: {population_std:.4f}")

# --------------------------------------------------
# PROVIDED FUNCTION: Calculate sample means
# --------------------------------------------------

def calculate_sample_means(data, sample_size, n_samples=1000):

    sample_means = []

    for i in range(n_samples):

        sample = np.random.choice(
            data,
            size=sample_size,
            replace=True
        )

        sample_means.append(np.mean(sample))

    return np.array(sample_means)


# TASK 2: Generate sample means for sample size 5
means_5 = calculate_sample_means(population, 5)

# TASK 3: Generate sample means for sample size 30
means_30 = calculate_sample_means(population, 30)

# TASK 4: Generate sample means for sample size 100
means_100 = calculate_sample_means(population, 100)

# TASK 5: Calculate standard deviation of each sampling distribution
std_5 = np.std(means_5)
std_30 = np.std(means_30)
std_100 = np.std(means_100)

print(f"\nSample Size 5: Std = {std_5:.4f}")
print(f"Sample Size 30: Std = {std_30:.4f}")
print(f"Sample Size 100: Std = {std_100:.4f}")

# TASK 6: Create histograms comparing sampling distributions
plt.figure(figsize=(10, 6))

plt.hist(
    means_5,
    bins=30,
    alpha=0.5,
    density=True,
    label="Sample Size = 5"
)

plt.hist(
    means_30,
    bins=30,
    alpha=0.5,
    density=True,
    label="Sample Size = 30"
)

plt.hist(
    means_100,
    bins=30,
    alpha=0.5,
    density=True,
    label="Sample Size = 100"
)

# TASK 7: Display results
plt.axvline(
    population_mean,
    color="black",
    linestyle="--",
    label="Population Mean"
)

plt.xlabel("Sample Mean")
plt.ylabel("Density")
plt.title("Central Limit Theorem: Sampling Distributions")

plt.legend()
plt.show()
```
