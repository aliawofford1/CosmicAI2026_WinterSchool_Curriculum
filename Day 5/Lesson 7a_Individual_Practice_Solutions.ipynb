Solutions

4. $$P = \frac{10}{2^n} = \frac{10}{2^5} = \frac{10}{32} = 0.3125 $$

## Individual Exercise 3:


```python
import numpy as np

np.random.seed(42)

even_outcomes = 3
total_outcomes = 6

rolls = np.random.randint(1, 7, size=1000)

# TASK 1
theoretical_probability = even_outcomes / total_outcomes

# TASK 2
even_count = np.sum(rolls % 2 == 0)

# TASK 3
experimental_probability = even_count / len(rolls)

# TASK 4
print(f"Theoretical Probability: {theoretical_probability:.3f}")
print(f"Experimental Probability: {experimental_probability:.3f}")
```

Answer: The theoretical probability of rolling an even number is 0.500 (50%). Using the simulation of 1,000 die rolls with np.random.seed(42), the experimental probability is 0.493 (49.3%).

The experimental probability differs from the theoretical probability by only 0.007 (0.7 percentage points), demonstrating that the simulation produces results close to the expected probability.


Question 2: Would you expect the experimental probability to become closer to the theoretical probability if you increased the simulation to 10,000 rolls? Explain.
Answer: Yes. Increasing the number of rolls from 1,000 to 10,000 would generally cause the experimental probability to become closer to the theoretical probability of 0.5.

With more trials, random fluctuations tend to have a smaller effect on the overall proportion of even outcomes. However, the experimental probability is not guaranteed to be closer in every individual simulation.

## Individual Exercise 4:


```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt

instrument_A = np.array([
    19, 20, 21, 20, 19,
    21, 20, 20, 19, 21
])

instrument_B = np.array([
    12, 25, 18, 27, 15,
    23, 17, 22, 16, 25
])

# Instrument A
mean_A = np.mean(instrument_A)
variance_A = np.var(instrument_A)
std_A = np.std(instrument_A)
sem_A = st.sem(instrument_A, ddof=0)

# Instrument B
mean_B = np.mean(instrument_B)
variance_B = np.var(instrument_B)
std_B = np.std(instrument_B)
sem_B = st.sem(instrument_B, ddof=0)

# Display results
print("Instrument A:")
print(f"Mean: {mean_A:.2f}")
print(f"Variance: {variance_A:.2f}")
print(f"Standard Deviation: {std_A:.2f}")
print(f"Standard Error: {sem_A:.2f}")

print("\nInstrument B:")
print(f"Mean: {mean_B:.2f}")
print(f"Variance: {variance_B:.2f}")
print(f"Standard Deviation: {std_B:.2f}")
print(f"Standard Error: {sem_B:.2f}")

# Visualize measurements
plt.figure(figsize=(8, 5))

plt.hist(
    instrument_A,
    bins=np.arange(10, 30, 2),
    alpha=0.6,
    label="Instrument A"
)

plt.hist(
    instrument_B,
    bins=np.arange(10, 30, 2),
    alpha=0.6,
    label="Instrument B"
)

plt.xlabel("Measurement Value")
plt.ylabel("Frequency")
plt.title("Comparison of Scientific Measurements")

plt.legend()
plt.show()
```

## Reflection Questions

### 1. Which instrument has the larger variance and standard deviation?

**Answer:** Instrument B has the larger variance and standard deviation.

- **Instrument A:** Variance = 0.60; Standard Deviation = 0.77

This indicates that Instrument B's measurements are more spread out around the mean compared to Instrument A.

### 2. What does a larger standard deviation indicate about the measurements?

**Answer:** A larger standard deviation indicates greater variability in the measurements. This means that individual measurements tend to be farther from the mean, suggesting less consistency in the recorded values.

### 3. How does the standard error differ from the standard deviation?

**Answer:** Standard deviation measures the variability of individual measurements around the mean, while standard error measures the uncertainty in the estimated mean.

Standard error is calculated as:

$$
SEM = \frac{\sigma}{\sqrt{n}}
$$

where:

- $\sigma$ is the standard deviation.
- $n$ is the number of measurements.

### 4. If both instruments measure the same quantity, which would you choose based on consistency alone? Explain.

**Answer:** Instrument A would be the better choice because it has a smaller variance and standard deviation. Its measurements are more closely grouped around the mean, indicating greater consistency.

However, consistency does not necessarily indicate accuracy. Determining which instrument is more accurate would require comparing its measurements against a known reference value.

### 5. If you collected more measurements, how would you expect the standard error to change?

**Answer:** The standard error would generally decrease as the number of measurements increases, assuming the variability of the measurements remains approximately constant.

Because standard error is inversely proportional to the square root of the sample size, collecting more measurements improves the precision of the estimated mean.

For example, increasing the sample size from 10 to 40 measurements would reduce the standard error by approximately half, assuming the standard deviation remains unchanged.
