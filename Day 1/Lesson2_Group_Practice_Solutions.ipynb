Objective: Use NumPy arrays, Boolean filtering, and statistical operations to identify exoplanets within a specified temperature range.

Scenario: Your research team has received a dataset containing estimated equilibrium temperatures for 50 hypothetical exoplanets. Your team wants to identify planets with temperatures that warrant further investigation.

**Group Instructions**

1. Load `exoplanet_temperatures.csv` using pandas.
2. Convert the temperature column into a NumPy array.
3. Convert the temperatures from Kelvin to Celsius.
4. Identify the planets with temperatures between 0°C and 50°C, inclusive.
5. Calculate the number of planets meeting this criterion and their average temperature.

**Group Discussion:** Does an equilibrium temperature between 0°C and 50°C guarantee that a planet can support liquid water on its surface? Why or why not?

```python

import numpy as np
import pandas as pd

df = pd.read_csv("exoplanet_temperatures.csv")

temperatures = df["Temperature_K"].to_numpy()
planets = df["Planet"].to_numpy()

temp_c = temperatures - 273.15

mask = (temp_c >= 0) & (temp_c <= 50)

selected_planets = planets[mask]
selected_temps = temp_c[mask]

print("Qualifying planets:", selected_planets)
print("Number of planets:", len(selected_planets))
print("Average temperature (°C):", np.mean(selected_temps))
```

Expected results
- 13 qualifying planets
- Mean temperature: approximately 25.61°C
- Qualifying IDs: EP-001, EP-006, EP-007, EP-009, EP-020, EP-025, EP-026, EP-028, EP-031, EP-033, EP-034, EP-035, EP-050

Discussion answer: No. Equilibrium temperature does not fully account for atmospheric greenhouse warming, surface pressure, or other conditions necessary for liquid water.

## Exercise 4

Objective: Use NumPy to analyze stellar brightness measurements and identify possible transit signals.

Scenario: Astronomers monitor stellar brightness to detect potential exoplanet transits. Your team has received 50 brightness measurements of a hypothetical star and must identify unusually low measurements.

**Group Instructions**

1. Load `stellar_brightness.csv` using pandas.
2. Convert the brightness measurements into a NumPy array.
3. Calculate the mean and minimum brightness.
4. Identify the observation number associated with the minimum brightness.
5. Count the number of observations with brightness below 0.99.
6. Calculate the percentage decrease from a normalized baseline brightness of 1.00 to the minimum brightness.

**Group Discussion:** Do the brightness measurements suggest a possible exoplanet transit? What additional evidence would help confirm the presence of an exoplanet?

```python

import numpy as np
import pandas as pd

df = pd.read_csv("stellar_brightness.csv")

brightness = df["Brightness"].to_numpy()
observations = df["Observation"].to_numpy()

mean_brightness = np.mean(brightness)
minimum = np.min(brightness)

minimum_index = np.argmin(brightness)
minimum_observation = observations[minimum_index]

low_count = np.sum(brightness < 0.99)

decrease = (1.00 - minimum) * 100

print("Mean brightness:", mean_brightness)
print("Minimum brightness:", minimum)
print("Observation of minimum:", minimum_observation)
print("Observations below 0.99:", low_count)
print("Brightness decrease (%):", decrease)
```

Expected results
- Mean brightness: 0.992258
- Minimum brightness: 0.9485
- Minimum occurs at observation 21
- 10 observations below 0.99
- Maximum brightness decrease: 5.15%


Discussion answer: The data contain two transit-like dips around observations 19–23 and 40–44. Repeated observations and evidence of a consistent orbital period would help assess whether these are planetary transits rather than stellar variability or instrumental effects.

## Exercise 6
Objective: Combine NumPy calculations with pandas DataFrame operations to analyze galaxy redshifts and approximate recession velocities.

Scenario: Your team has obtained redshift measurements for 50 hypothetical galaxies. You have been asked to estimate their recession velocities and identify the galaxies receding fastest.

For small cosmological redshifts, approximate recession velocity using:

$$
v \approx cz
$$

where:
- $v$ is the recession velocity (km/s)
- $c = 300{,}000$ km/s is the approximate speed of light
- $z$ is the galaxy's redshift.


**Group Instructions**

1. Load `galaxy_redshifts.csv` into a pandas DataFrame.
2. Convert the redshift values into a NumPy array.
3. Calculate the approximate recession velocity for each galaxy.
4. Add the calculated velocities to the DataFrame in a new column named `Velocity_km_s`.
5. Identify galaxies with recession velocities greater than 6,000 km/s.
6. Determine which galaxy has the highest recession velocity and report its value.

**Group Discussion:** What relationship do you observe between galaxy redshift and recession velocity? How can astronomers use redshift measurements to investigate the expansion of the universe?

```python

import numpy as np
import pandas as pd

df = pd.read_csv("galaxy_redshifts.csv")

c = 300000

# Convert redshifts to NumPy array
redshifts = df["Redshift"].to_numpy()

# Calculate recession velocities
velocities = redshifts * c

# Add a new DataFrame column
df["Velocity_km_s"] = velocities

# Filter galaxies
fast = df[df["Velocity_km_s"] > 6000]

print("Galaxies above 6000 km/s:")
print(fast)

print("Number of galaxies:", len(fast))

# Identify the galaxy with the highest velocity
fastest = df.loc[df["Velocity_km_s"].idxmax()]

print("Galaxy with highest velocity:", fastest["Galaxy"])
print("Highest velocity (km/s):", fastest["Velocity_km_s"])
```

Expected results
- 39 galaxies exceed 6,000 km/s.
- Galaxy G-009 has the highest recession velocity.
- Maximum velocity: 17,907 km/s.
- Corresponding redshift: 0.05969.

Discussion answer: Under the low-redshift approximation, recession velocity increases proportionally with redshift. Astronomers use cosmological redshift measurements, together with distance information and cosmological models, to study the expansion of the universe.
