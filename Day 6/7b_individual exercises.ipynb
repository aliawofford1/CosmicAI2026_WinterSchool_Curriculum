You will use logistic regression to predict a binary outcome:

- 0 = No genuine astronomical signal
- 1 = Genuine astronomical signal

Unlike linear regression, logistic regression models the probability of belonging to a category.

For a single predictor:

$$
P(Y=1 \mid X=x) =
\frac{1}{1+e^{-(b_0+b_1x)}}
$$

The predicted probability can be converted into a class label using a classification threshold.

- Predict class 0 when $P(Y=1) < 0.5$.
